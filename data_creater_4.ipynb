```python
import cv2
import numpy as np
import os
import glob

def manually_crop_first_image(img_input):
    """
    Allows the user to manually mask and crop the first image.
    Returns the cropped template and mask.
    """
    global drawing, points, img, img_copy
    drawing = False
    points = []
    img = img_input
    img_copy = img.copy()

    def draw_freehand(event, x, y, flags, param):
        global drawing, points, img_copy

        if event == cv2.EVENT_LBUTTONDOWN:
            drawing = True
            points = [(x, y)]
            img_copy = img.copy()

        elif event == cv2.EVENT_MOUSEMOVE and drawing:
            img_copy = img.copy()
            points.append((x, y))
            for i in range(1, len(points)):
                cv2.line(img_copy, points[i - 1], points[i], (0, 255, 0), 2)
            cv2.imshow("Image", img_copy)

        elif event == cv2.EVENT_LBUTTONUP:
            drawing = False
            points.append((x, y))
            cv2.line(img_copy, points[0], points[-1], (0, 255, 0), 2)
            cv2.imshow("Image", img_copy)

    cv2.imshow("Image", img)
    cv2.setMouseCallback("Image", draw_freehand)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

    # Create a mask and crop the template
    mask = np.zeros(img.shape[:2], dtype=np.uint8)
    if len(points) > 2:
        points_array = np.array(points, dtype=np.int32)
        cv2.fillPoly(mask, [points_array], 255)

        # Crop the region of interest
        x, y, w, h = cv2.boundingRect(points_array)
        template = img[y:y+h, x:x+w]
        mask_cropped = mask[y:y+h, x:x+w]

        return template, mask_cropped
    else:
        raise ValueError("No valid mask created. Draw a complete polygon.")

def find_and_crop_template(img, template, mask):
    """
    Finds the region in the image that matches the template and crops it.
    """
    # Perform template matching
    result = cv2.matchTemplate(img, template, cv2.TM_CCOEFF_NORMED, mask=mask)
    _, max_val, _, max_loc = cv2.minMaxLoc(result)

    if max_val < 0.6:  # Threshold for matching confidence
        print("No matching region found with high confidence.")
        return None

    # Crop the matched region
    h, w = template.shape[:2]
    x, y = max_loc
    cropped_img = img[y:y+h, x:x+w]
    return cropped_img

def process_images(image_folder, save_dir, template, mask):
    """
    Processes all images in the folder, finding and cropping the template region.
    """
    # Get all image files in the folder
    image_files = glob.glob(os.path.join(image_folder, "*.jpg")) + \
                  glob.glob(os.path.join(image_folder, "*.png")) + \
                  glob.glob(os.path.join(image_folder, "*.jpeg"))

    if not image_files:
        print("No images found in the specified folder.")
        return

    # Create save directory if it doesn't exist
    os.makedirs(save_dir, exist_ok=True)

    # Process all images
    for image_path in image_files:
        print(f"Processing image: {image_path}")
        img = cv2.imread(image_path)
        if img is not None:
            cropped_img = find_and_crop_template(img, template, mask)
            if cropped_img is not None:
                save_path = os.path.join(save_dir, os.path.basename(image_path))
                cv2.imwrite(save_path, cropped_img)
                print(f"Saved cropped image to: {save_path}")
            else:
                print(f"No match found for: {image_path}")
        else:
            print(f"Failed to load image: {image_path}")

# Example usage
# Manually mask and crop the first image
first_image_path = '/home/natsu/MechAI/Dataset/disk_rotor/disk_rotor_1.png'
first_image = cv2.imread(first_image_path)
if first_image is None:
    print(f"Failed to load the first image: {first_image_path}")
else:
    print("Draw the region to crop on the first image.")
    template, mask = manually_crop_first_image(first_image)

    # Automatically process remaining images
    process_images(
        '/home/natsu/MechAI/Dataset/disk_rotor',
        '/home/natsu/MechAI/Dataset/disk_rotor_cropped',
        template,
        mask
    )
```
